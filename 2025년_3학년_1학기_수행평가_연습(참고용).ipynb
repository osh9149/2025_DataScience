#문제5.지역별로 대학교의 개수를 카운트하여 데이터를 내림차순으로 정렬하여 새로운 데이터프레임(df1)으로 만든다.(4점)

#문제6. 데이터프레임(d1f)에서 대학교 개수가 가장 많은 지역과 가장 적은 지역을 찾아 다음과 같이 출력한다.(4점)

```python
#문제6번(4점)

print('전국에서 대학교의 수가 가장 많은 지역은 %s입니다.'%df1['학교'].index[0])
print('전국에서 대학교의 수가 가장 많은 지역은 %s입니다.'%df1['학교'].idxmax())

print('전국에서 대학교의 수가 가장 적은 지역은 %s입니다.'%df1['학교'].index[-1])
print('전국에서 대학교의 수가 가장 적은 지역은 %s입니다.'%df1['학교'].idxmin())



```

#문제13. 데이터프레임(df2)을 그림처럼 그룹화(평균을 계산)하여 새로운 데이터프레임(df3)를 만든 후 의학을 기준으로 내림차순으로 정렬하여 출력한다. (4점)
image.png

```python
#문제13번(4점)

#groupby('컬럼').count() : 개수
#groupby('컬럼').sum() : 합계
#groupby('컬럼').mean() : 평균

df3=df2.groupby(df2['설립구분'])['의학'].mean()
df3
```

#문제15. 데이터프레임(df2)에서 원하는 지역의 이름을 입력받아 해당하는 지역의 데이터만 새로운 데이터프레임(df4)로 만든다.(4점)
예) '서울' 입력하여 확인

```python
#문제15번(4점)

name=input('지역을 입력하세요 : ')
df4=df2[df2['지역'].str.contains(name, na=False)]
df4




```

#문제16. 데이터프레임(df4)에서 위도 경도를 찾아 지도에 마커로 표시(zoom_start=11)하고 팝업에 학교이름이 보이도록 한다.(4점)
아이콘 변경 -> icon=folium.Icon(icon = 'home', color='red')

20230614_085949.png

```python
#문제16번(4점)

lat = df4.loc[df4.index[0], '위도']
lng = df4.loc[df4.index[0], '경도']

map = folium.Map(
    location=[lat,lng ], tiles ='OpenStreetMap' , zoom_start=11)

for idx in df4.index:

    lat = df4.loc[idx, '위도']
    lng = df4.loc[idx, '경도']
    name = df4.loc[idx, '학교']

    folium.Marker(
        location=[lat, lng],
        icon=folium.Icon(color='red'),
				popup=name
    ).add_to(map)

map

```
