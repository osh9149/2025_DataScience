#loc(location) 데이터프레임의 특정 행이나 컬럼을 추출해오는 방법

```python
#전체 데이터 프레임에서 인덱스이름이 '0'인행만 추출

df.loc[0]
```

```python
df.loc[0,'학교']
```

```python
df.loc[:,'학교']
```

```python
df.loc[:10,'학교':'등록금']
```

#iloc(interger location) 데이터프레임의 행이나 컬럼의 순서를 나타내는 정수로 특정 값을 추출해오는 방법

```python
#전체 데이터 프레임에서 0번째 행에 있는 값들만 추출

df.iloc[0]
```

```python
#df.iloc[행인덱스, 열인덱스]

df.iloc[0,2]
```

```python
df.iloc[:5, :5]
```
